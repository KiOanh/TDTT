[Rounding]
Nhập vào một số thực: in ra kết quả làm tròn lên, tròn xuống, làm tròn tới số
nguyên gần nhất (không được sử dụng hàm sẵn có của pyton). Các kết quả
cách nhau một dấu cách

```python
a=float(input())
tronxuong=tronlen=gannhat=int(a)
thapphan=a-float(tronxuong)
if a<0 and thapphan:
    tronxuong-=1
    tronlen-=1
    gannhat-=1
    thapphan*=-1
if thapphan:
   tronlen+=1
if thapphan>=0.5:
   gannhat+=1
print(tronxuong,tronlen,gannhat)
```
